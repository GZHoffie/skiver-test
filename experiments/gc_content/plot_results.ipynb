```python
import matplotlib.pyplot as plt
import numpy as np

import pandas as pd

align_df = pd.read_csv("../../output/zymo/SRR7498042_assembly.per_aln_stats.csv")
align_df.head()


#plt.plot(align_df["gc_content"], align_df["concordance"], 'o', alpha=0.01, markersize=1)
gc_bins = pd.cut(align_df["gc_content"], bins=np.arange(0, 1.1, 0.05))
mean_concordance = align_df.groupby(gc_bins)["concordance"].mean()

plt.plot(mean_concordance.index.categories.mid, mean_concordance, marker='o')
plt.xlabel("GC Content")
plt.ylabel("Mean Concordance")
plt.title("Mean Concordance vs. GC Content")
plt.grid(True)
plt.show()
```
