```python
import matplotlib.pyplot as plt
import numpy as np

import pandas as pd

align_df = pd.read_csv("../../output/zymo/SRR7498042_assembly.per_aln_stats.csv")
#align_df.head()

skiver_df = pd.read_csv("../../output/zymo/skiver/SRR7498042_ref.summary_gc_content.csv")
skiver_df = skiver_df[skiver_df["num_correct"] + skiver_df["beta"] > 5000]
#skiver_df.head()


#plt.plot(align_df["gc_content"], align_df["concordance"], 'o', alpha=0.01, markersize=1)
align_df["gap_compressed_error_rate"] = 1 - align_df["gap_compressed_concordance"]
gc_bins = pd.cut(align_df["gc_content"] , bins=np.arange(0, 1.1, 0.01))
mean_concordance = align_df.groupby(gc_bins)["gap_compressed_error_rate"].mean()

plt.Figure(figsize=(10, 6))
plt.plot(skiver_df["gc_content_min"] / 100, skiver_df["per_base_error_rate"], marker='o', label='Skiver')
plt.plot(mean_concordance.index.categories.mid, mean_concordance, marker='o')
plt.xlabel("GC Content")
plt.ylabel("Mean Concordance")
plt.title("Mean Concordance vs. GC Content")
plt.grid(True)
plt.show()
```
